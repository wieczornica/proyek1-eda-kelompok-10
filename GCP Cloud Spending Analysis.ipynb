```python
ringkasan_biaya = df.groupby("Service Name")["Rounded Cost ($)"].agg(
    rata2="mean",
    std="std",
    jumlah_data="count"
)

ringkasan_biaya["CV (%)"] = (ringkasan_biaya["std"] / ringkasan_biaya["rata2"]) * 100
ringkasan_biaya = ringkasan_biaya.round(2)
ringkasan_biaya
```

```python
rata2_cpu = df["CPU Utilization (%)"].mean()
std_cpu = df["CPU Utilization (%)"].std()
cv_cpu = (std_cpu / rata2_cpu) * 100


print(f"Rata-rata CPU Utilization: {rata2_cpu:.2f}%")
print(f"Standar deviasi CPU Utilization: {std_cpu:.2f}%")
print(f"Koefesien variasinya: {cv_cpu:.2f}%")
```
