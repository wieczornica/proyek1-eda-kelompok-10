```python
print('Variabel Numerik')
hasil = pd.DataFrame({
    "Usage Quantity": [
        df["Usage Quantity"].mean(),
        df["Usage Quantity"].median(),
        df["Usage Quantity"].mode()[0],
        df["Usage Quantity"].var(),
        df["Usage Quantity"].std(),
        df["Usage Quantity"].quantile(0.25),
        df["Usage Quantity"].quantile(0.75)
    ],
    
    "CPU Utilization (%)": [
        df["CPU Utilization (%)"].mean(),
        df["CPU Utilization (%)"].median(),
        df["CPU Utilization (%)"].mode()[0],
        df["CPU Utilization (%)"].var(),
        df["CPU Utilization (%)"].std(),
        df["CPU Utilization (%)"].quantile(0.25),
        df["CPU Utilization (%)"].quantile(0.75)
    ],
    
    "Total Cost (INR)": [
        df["Total Cost (INR)"].mean(),
        df["Total Cost (INR)"].median(),
        df["Total Cost (INR)"].mode()[0],
        df["Total Cost (INR)"].var(),
        df["Total Cost (INR)"].std(),
        df["Total Cost (INR)"].quantile(0.25),
        df["Total Cost (INR)"].quantile(0.75)
    ]
}, index=[
    "Mean",
    "Median",
    "Mode",
    "Variance",
    "Standard Deviation",
    "Q1",
    "Q3"
])

hasil #Nilai plus: hitung varians atau simpangan baku untuk satu kolom secara manual dengan rumus (misalnya memakai 10 baris pertama), lalu cocokkan dengan hasil pandas. Perhatikan bahwa var() dan std() di pandas memakai pembagi n − 1.
```
